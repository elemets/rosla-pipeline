```python
def clean_demographics(df):
    """Cleans Gender and Race columns by normalizing case and mapping outliers."""
    # Normalize Gender to title case and fix 'F'
    df['Gender'] = df['Gender'].str.title().replace({'F': 'Female'})
    
    # Filter to only allowed Gender values, everything else becomes 'Unknown' or NaN
    valid_genders = ['Male', 'Female', 'Non-Binary']
    df.loc[~df['Gender'].isin(valid_genders), 'Gender'] = 'Unknown'
    
    # Clean Race: Convert 'k' to 'UNKNOWN'
    df['Race'] = df['Race'].replace({'k': 'UNKNOWN'})
    
    return df
```

```python
latest_df = clean_demographics(latest_df)
```
