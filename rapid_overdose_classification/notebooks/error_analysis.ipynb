```python
import pandas as pd
import numpy as np
```

```python
desert_rate_ana = pd.read_csv("../../data/OD_ZIP_2024_Q4.CSV")
```

```python
def convert_to_num(row):
    if row == "SUPPRESSED":
        return np.nan
    else:
        return float(row)
```

```python
desert_rate_ana["Rate"] = desert_rate_ana["Rate"].apply(convert_to_num)
```

```python
desert_rate_ana.to_csv("../../reports/OD_ZIP_Cleaned.csv")
```
