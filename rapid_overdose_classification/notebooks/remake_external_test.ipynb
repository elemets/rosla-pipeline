```python
import pandas as pd
```

```python
misclass = pd.read_csv(
    "../../data/misclassifications_1_23_25_recoded_ip.csv", index_col="Unnamed: 0"
)
```

```python
for index, row in misclass.iterrows():
    if (
        "ethanolism".lower() in row["text"].lower()
        or "ethanol" in row["text"].lower()
        or "alcohol" in row["text"].lower()
        or "alcoholism" in row["text"].lower()
    ):
        misclass.at[index, "Alcohol"] = 1
        misclass.at[index, "Any Drugs"] = 1
```

```python
misclass.to_csv("../../data/misclassifications_1_23_25_recoded.csv")
```

```python
ext_test = pd.read_csv("../../data/external_test.csv", index_col="Unnamed: 0")
```

```python
ext_test.update(misclass)
```

```python
ext_test.to_csv("../../data/recoded_ext_test2801_recode.csv")
```
