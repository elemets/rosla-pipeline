```python
import pandas as pd
```

```python
overdoses = pd.read_csv("../../pipeline/pipeline_steps/input_files/2012-01-2025-01-overdoses.csv")
```

```python
pd.merge(overdoses['Race'], overdoses['Races'])
```

```python
import numpy as np
def clean_and_categorize_race(race):
    if pd.isna(race):
        return np.nan
    race = race.lower()  # Case folding
    race = race.replace(" ", "")  # Remove spaces
    race = race.replace("\n", "")  # Remove newline characters
    if '","' in race or "," in race:  # Adjust based on your actual separator
        return "UNKNOWN"

    if race in ["americanindian", "nativeamerican", 'footcellulitisnativeamerican']:
        return "AMERICAN INDIAN"
    elif race in ["armenian", "middleeastern", 'middleeasternornorth']:
        return "MIDDLE EASTERN"
    elif race in [
        "asian",
        "cambodian",
        "chinese",
        "filipino",
        "japanese",
        "korean",
        "eastindian",
        "thai",
        "vietnamese",
    ]:
        return "ASIAN"
    elif race in ["black"]:
        return "BLACK"
    elif race in [
        "guamanian",
        "hawaiian",
        "pacificislander",
        "samoan",
        "tongan",
        "nativehawaiian/otherpacificislander",
        'islander',
    ]:
        return "PACIFIC ISLANDER"
    elif race in ["hispanic/latino", "hispanic/latina", "hispanic/latinamerican", "latine"]:
        return "LATINE"
    elif race in ["white", "caucasian", "white/caucasian"]:
        return "WHITE"
    elif race in ["unknown", "null", "unknown/other", 'cardiovasculardiseasenull']:
        return np.nan
    else:
        return race
```

```python
overdoses['Race'] = overdoses['Race'].fillna(overdoses['Races'])
```

```python
overdoses['Race'] = overdoses['Race'].apply(clean_and_categorize_race)
```

```python
overdoses['Race'].value_counts()
```

```python
overdoses = overdoses.drop(columns=['Races'])
```

```python
overdoses.to_csv("../../pipeline/pipeline_steps/input_files/2012-01-2025-01-overdoses.csv", index=False)
```
